```python
df = spark.read.table('medical_catalog.bronze.payers')
display(df.limit(10))
```

```python
#Convert columns to snakecase
for col in df.columns:
    new_col = col.lower().replace(' ', '_')
    if new_col != col:
        df = df.withColumnRenamed(col, new_col)
display(df.limit(10))
```

```python
# Fill nulls in address columns, city, and state_headquarted with 'Unknown'
address_cols = [col for col in df.columns if 'address' in col]
extra_cols = ['city', 'state_headquartered',]
fill_cols = address_cols + [col for col in extra_cols if col in df.columns]
df = df.fillna('Unknown', subset=fill_cols)
df = df.fillna(000,subset=['zip'])

# Drop phone_number column if it exists
df = df.drop('phone')

display(df.limit(10))
```

```python
# Writing Data in Silver Table
df.write.mode("overwrite").saveAsTable("medical_catalog.silver.payers")
```
