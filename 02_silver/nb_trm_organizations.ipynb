```python
df = spark.read.table('medical_catalog.bronze.organizations')
display(df.limit(10))
```

```python
# Convert columns to snakecase
for col in df.columns:
    new_col = col.lower().replace(' ', '_')
    if new_col != col:
        df = df.withColumnRenamed(col, new_col)
```

```python
df.write.mode("overwrite").saveAsTable("medical_catalog.silver.organizations")
```
