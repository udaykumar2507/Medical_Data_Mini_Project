```sql
CREATE OR REPLACE TABLE medical_catalog.gold.dim_calendar AS
SELECT 
  date,
  year(date) as year,
  month(date) as month,
  date_format(date, 'MMMM') as month_name,
  concat('Q', quarter(date)) as quarter,

  CASE 
    WHEN month(date) <= 6 THEN 'H1'
    ELSE 'H2'
  END as half_year
FROM (
  SELECT explode(sequence(
    to_date('2020-01-01'),
    to_date('2030-12-31'),
    interval 1 day
  )) as date
);






```
