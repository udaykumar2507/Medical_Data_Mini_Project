### Fact Table 

```sql
CREATE OR REPLACE TABLE medical_catalog.gold.fact AS
SELECT 
    e.id as encounter_id,
    e.patient as patient_id ,
    e.payer as payer_id,
    e.organization as organization_id,
    e.encounter_class,
    e.start as start_time,
    e.stop as end_time,

    -- Derived column
    (unix_timestamp(e.stop) - unix_timestamp(e.start)) / 3600 
      AS duration_hours,

    e.base_encounter_cost as base_cost,
    e.total_claim_cost

FROM medical_catalog.silver.encounters e;
```
