### KPI - PATIENT STAY TIME (Under 24 hrs vs Over 24 _hrs_)

```sql

SELECT 
  fc.year,
  fc.month,

  CASE 
    WHEN fc.duration_hours > 24 THEN 'Over 24 hrs'
    ELSE 'Under 24 hrs'
  END as type,

  COUNT(DISTINCT encounter_id) as total,

  ROUND(
    COUNT(DISTINCT encounter_id) * 100.0 / 
    SUM(COUNT(DISTINCT encounter_id)) OVER (PARTITION BY fc.year, fc.month),
    2
  ) as percentage

FROM medical_catalog.gold.finalcube fc

GROUP BY fc.year, fc.month, type
ORDER BY fc.year, fc.month;
```
