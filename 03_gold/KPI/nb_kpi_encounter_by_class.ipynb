#### Total Number of Encounters By Year,Quarter and by Encounter Class


```sql

SELECT 
  fc.year,
  fc.quarter,
  fc.encounter_class,

  COUNT(DISTINCT fc.encounter_id) as total,

  ROUND(
    COUNT(DISTINCT fc.encounter_id) * 100.0 / 
    SUM(COUNT(DISTINCT fc.encounter_id)) OVER (PARTITION BY fc.year, fc.quarter),
    2
  ) as percentage

FROM medical_catalog.gold.finalcube fc
GROUP BY fc.year, fc.quarter, fc.encounter_class
ORDER BY fc.year, fc.quarter;
```
