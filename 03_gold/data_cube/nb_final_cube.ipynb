### DATACUBE (MERGE FACT,PAYER,PROCEDURE,CALENDER)

```sql
CREATE OR REPLACE TABLE medical_catalog.gold.finalcube AS
SELECT 
    f.encounter_id,
    f.patient_id,
    f.payer_id,
    f.encounter_class,
    f.duration_hours,
    f.base_cost,
    f.total_claim_cost,
    f.start_time,
    f.end_time,
    c.year,
    c.month,
    c.quarter,
    c.half_year,
    COALESCE(p.payer_name, 'No Insurance') as payer_name,
    pr.procedure_code

FROM medical_catalog.gold.fact f

LEFT JOIN medical_catalog.gold.dim_calendar c 
  ON DATE(f.start_time) = c.date

LEFT JOIN medical_catalog.gold.dim_payer p 
  ON f.payer_id = p.payer_id

LEFT JOIN medical_catalog.gold.dim_procedure pr 
  ON f.encounter_id = pr.encounter_id;
```
