```python
files = [
    "encounters.csv",
    "procedures.csv",
    "payers.csv",
    "patients.csv",
    "organizations.csv"
]

for file in files:
    df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(
        f"s3://medicalrawdata-de/{file}"
    )
    # Rename columns to replace spaces and invalid characters
    for col in df.columns:
        new_col = col.replace(" ", "_")
        if new_col != col:
            df = df.withColumnRenamed(col, new_col)
    table_name = file.replace(".csv", "")
    target = f"medical_catalog.bronze.{table_name}"
    df.write.mode("overwrite").saveAsTable(target)
```
